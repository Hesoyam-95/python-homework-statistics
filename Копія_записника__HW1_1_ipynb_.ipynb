**Завдання 2.**

```python
sum_purchases = [ 316,  493,  367,  153,  224,  121,  322,  450,  196,  397,  281,
  239,  484,  124,  223,  448,  432,  255, 4310, 3524]

total = sum(sum_purchases)
count = len(sum_purchases)
mean_python = total / count
print("Середнє (Python):", mean_python)

mean_numpy = np.mean(sum_purchases)
print("Середнє (NumPy):", mean_numpy)
```

```python
sorted_purchases = sorted(sum_purchases)
print(sorted_purchases)
n = len(sorted_purchases)
print(n)
first = sorted_purchases[9]
second = sorted_purchases[10]
print(first)
print(second)
median_python = (first + second) / 2
print("Медіана:", median_python)
#Вибіркове середнє використовую для визначення загального середнього значення, якщо немає значних відхилень у даних.
```

```python
median_numpy = np.median(sum_purchases)

print("Медіана (NumPy):", median_numpy)
#Медіану — коли є дуже великі або малі значення, які спотворюють середнє. Наприклад, для аналізу зарплат або цін.
```

**Завдання 4**

```python
mean = sum(sum_purchases) / len(sum_purchases)
n = len(sum_purchases)
squared_differences = sum((x - mean)**2 for x in sum_purchases)
std_python = (squared_differences / (n - 1))**0.5

print("Стандартне відхилення (Python):", round(std_python, -1))
```

```python
std_numpy = np.std(sum_purchases, ddof=1)

print("Стандартне відхилення (NumPy):", round(std_numpy, -1))
```

**Завдання 7**

```python
sum_purchases = np.array([
    316, 493, 367, 153, 224, 121, 322, 450, 196, 397,
    281, 239, 484, 124, 223, 448, 432, 255, 4310, 3524
])

visits = np.array([
    6, 2, 10, 3, 9, 9, 2, 3, 6, 6,
    10, 2, 5, 10, 3, 5, 4, 7, 4, 2
])

threshold = np.mean(sum_purchases) + 2 * np.std(sum_purchases, ddof=1)

mask = sum_purchases <= threshold

filtered_purchases = sum_purchases[mask]
filtered_visits = visits[mask]

correlation = np.corrcoef(filtered_purchases, filtered_visits)[0, 1]

print("Поріг:", threshold)
print("Коефіцієнт кореляції Пірсона:", correlation)

plt.scatter(filtered_visits, filtered_purchases)
plt.xlabel("Кількість відвідувань")
plt.ylabel("Сума покупок")
plt.title("Кореляція між відвідуваннями та сумою покупок")
plt.grid()
plt.show()
# Після видалення викидів кореляція трохи змінилася. Зв’язок між кількістю відвідувань і сумою покупок слабкий: більше відвідувань не означає більшу суму покупок.
```
